```python
import pandas as pd

def generate_signal(row, threshold):
    
    bull = row["Bull"]
    bear = row["Bear"]
    sideways = row["Sideways"]

    if bull > threshold :
        return "BUY"
    elif bear > threshold :
        return "SELL"
    elif sideways > threshold :
        return "HOLD"
    else :
        return "HOLD"
    

```

```python
def generate_positon(row):
    bull = row["Bull"]
    bear = row["Bear"]

    position = bull - bear
    return round(position, 2)
```
