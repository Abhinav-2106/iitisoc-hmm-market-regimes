```python
import pandas as pd

def generate_signal(row, threshold):
    
    bull = row["Bull"]
    bear = row["Bear"]
    sideways = row["Sideways"]

    if bull > threshold :
        return "BUY"
    elif bear > threshold :
        return "SELL"
    elif sideways > threshold :
        return "HOLD"
    else :
        return "HOLD"
    

```

```python
def generate_positon(row):
    bull = row["Bull"]
    bear = row["Bear"]

    position = bull - bear
    return round(position, 2)
```

```python
import pandas as pd
import os

path = os.path.realpath("analysis")
dir = os.path.dirname(path).replace("notebooks", "data")

os.chdir(dir)

df = pd.read_csv("state_probabilities.csv")
df
```

```python
import numpy as np

P = pd.read_csv("transition_matrix.csv").values

state_cols = [f"State_{i}" for i in range(6)]

def forecast(curr_regime_prob):
    return curr_regime_prob @ P

tomorrow_probs = pd.DataFrame(
    df[state_cols].apply(
        lambda row: forecast(row.values),
        axis=1
    ).tolist(),
    columns=state_cols
)

tomorrow_probs
```
